**2C.2** — the chained-replace bug, then the fix.

```python
dna = "ATGC"
broken = dna.replace("A","T").replace("T","A").replace("G","C").replace("C","G")
print("broken:", broken)
# Wrong because replace runs sequentially: A->T makes T's, then T->A turns them
# (and the original T) all back to A; G/C collide the same way.
correct = dna.translate(str.maketrans("ACGT","TGCA"))[::-1]
print("correct rev-comp:", correct)
```

**3C.1**

```python
seq = "ATGAAACCCTAA"
i = 0
while i < len(seq) - 2:
    codon = seq[i:i+3]
    if codon in ("TAA", "TAG", "TGA"):
        print("first stop at", i)
        break
    i += 3
```

**3C.2** — minimal ORF finder.

```python
seq = "CCCATGAAACCCGGGTAACC"
start = seq.find("ATG")
orf = ""
i = start
while i < len(seq) - 2:
    codon = seq[i:i+3]
    orf += codon
    if codon in ("TAA", "TAG", "TGA"):
        break
    i += 3
print("ORF:", orf)
```

**Module 4 challenge** — base-composition report.

```python
seq = "ATGCGCGTATTACCGGGCAT"
if not set(seq) <= {"A","C","G","T"}:
    print("WARNING: invalid bases present")
counts = {}
for base in seq:
    counts[base] = counts.get(base, 0) + 1
gc = (counts.get("G",0) + counts.get("C",0)) / len(seq)
print(f"Length: {len(seq)}")
print(f"Valid:  {set(seq) <= set('ACGT')}")
print(f"Composition: {counts}")
print(f"GC content: {gc:.2%}")
```

**5B.2** — mutable default bug and fix.

```python
def collect_buggy(item, acc=[]):
    acc.append(item)
    return acc
print(collect_buggy("rpoB"))
print(collect_buggy("katG"))   # contains rpoB too — same list reused

def collect_fixed(item, acc=None):
    if acc is None:
        acc = []        # fresh list every call
    acc.append(item)
    return acc
print(collect_fixed("rpoB"))
print(collect_fixed("katG"))   # independent now
```

**Part I capstone** — full sequence report from your own functions.

```python
def gc_content(seq):
    seq = seq.upper()
    return (seq.count("G") + seq.count("C")) / len(seq)

def reverse_complement(seq):
    return seq.upper().translate(str.maketrans("ACGT", "TGCA"))[::-1]

def is_valid_dna(seq):
    return set(seq.upper()) <= {"A", "C", "G", "T"}

def base_composition(seq):
    counts = {}
    for base in seq.upper():
        counts[base] = counts.get(base, 0) + 1
    return counts

def codons(seq):
    return [seq[i:i+3] for i in range(0, len(seq) - 2, 3)]

def sequence_report(seq):
    seq = seq.upper()
    print(f"Length:       {len(seq)} bp")
    print(f"Valid DNA:    {is_valid_dna(seq)}")
    print(f"Composition:  {base_composition(seq)}")
    print(f"GC content:   {gc_content(seq):.2%}")
    print(f"Rev-comp:     {reverse_complement(seq)}")
    print(f"Codons:       {codons(seq)}")

sequence_report("ATGCGTACGGGCTAA")
```
