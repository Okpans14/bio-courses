## 5A. Basic functions
### Worked example

```python
def gc_content(seq):
    """Return the GC fraction of a DNA sequence."""
    seq = seq.upper()
    return (seq.count("G") + seq.count("C")) / len(seq)

print(gc_content("ATGCGC"))
help(gc_content)
```

### 5A.2 (harder)
Write `reverse_complement(seq)` (with docstring) using `translate` + `[::-1]`, handling lowercase input by upper-casing first. Test on `"atgc"` → `GCAT`.

```python
# your code here
def reverse_complement(seq):
    """Return the reverse complement of a DNA sequence"""
    seq = seq.upper()
    compl = seq.translate(str.maketrans("ACGT", "TGCA"))
    return compl[::-1]
reverse_complement("atgc")
```

## 5B. Arguments, scope, multiple returns
### Worked example

```python
def base_summary(seq, frame=0):
    """Returns the summary of a sequence in a given frame (default 0)."""
    seq = seq.upper()
    gc = (seq.count("G") + seq.count("C")) / len(seq)
    return len(seq), gc

length, gc = base_summary("ATGCGC") #unpack into two names
print(length, gc)
```

Write a single function `sequence_report(seq)` that takes a DNA string and returns (or prints) a full report:
